## 1. First Sales Calculator

This exercise calculates a rider's gross income, net income and profit margin using daily trip information.

```python
# Sales information
trips = 18
avg_fare = 220.0
fuel_cost = 1100.0
rider_name = "Kelvin"

# Calculations
gross_income = trips*avg_fare
net_income = gross_income-fuel_cost
margin_pct = (net_income/gross_income)*100

# Display Results
print(f"{rider_name}: Gross income: KES {gross_income:,.2f},")
f"Net income: KES {net_income:,.2f}, Profit margin: {margin_pct:.1f}%"
```

## 2. Inventory alerts with Dictionaries and loops.

This exercise identifies products whose stock levels are equal to or below their reorder levels.

```python
inventory = [
    {"name": "Unga", "stock": 12, "reorder_level": 15},
    {"name": "Sugar", "stock": 40, "reorder_level": 20},
    {"name": "Milk", "stock": 8, "reorder_level": 10},
    {"name": "Bread", "stock": 25, "reorder_level": 10},
]

def restock_report(items):
  to_reorder = []
  for item in items:
    if item["stock"] <= item["reorder_level"]:
      status = "REORDER"
      to_reorder.append(item["name"])
    else:
      status = "OK"
    print(f"{item['name']:<6} stock={item['stock']:>3} {status}")
  return to_reorder

print(restock_report(inventory))
```
