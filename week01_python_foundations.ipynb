## 1. First Sales Calculator

This exercise calculates a rider's gross income, net income and profit margin using daily trip information.

```python
# Sales information
trips = 18
avg_fare = 220.0
fuel_cost = 1100.0
rider_name = "Kelvin"

# Calculations
gross_income = trips*avg_fare
net_income = gross_income-fuel_cost
margin_pct = (net_income/gross_income)*100

# Display Results
print(f"{rider_name}: Gross income: KES {gross_income:,.2f},")
f"Net income: KES {net_income:,.2f}, Profit margin: {margin_pct:.1f}%"
```

## 2. Inventory alerts with Dictionaries and loops.

This exercise identifies products whose stock levels are equal to or below their reorder levels.

```python
inventory = [
    {"name": "Unga", "stock": 12, "reorder_level": 15},
    {"name": "Sugar", "stock": 40, "reorder_level": 20},
    {"name": "Milk", "stock": 8, "reorder_level": 10},
    {"name": "Bread", "stock": 25, "reorder_level": 10},
]

def restock_report(items):
  to_reorder = []
  for item in items:
    if item["stock"] <= item["reorder_level"]:
      status = "REORDER"
      to_reorder.append(item["name"])
    else:
      status = "OK"
    print(f"{item['name']:<6} stock={item['stock']:>3} {status}")
  return to_reorder

print(restock_report(inventory))
```

## 3. Working Example: Weekly Sales Report

**Business:** Mama Njeri's Kiosk  
**Location:** Kawangware  
**Period:** Monday to Sunday

### Business Questions by Mama Njeri
1. How much did I make this week?
2. What does a typical customer spend?
3. Which days should I focus on improving?

```python
week = [
    {"day":"Mon","sales":1200,"customers":30},
    {"day":"Tue","sales":950,"customers":25},
    {"day":"Wed","sales":1800,"customers":41},
    {"day":"Thu","sales":1430,"customers":33},
    {"day":"Fri","sales":2100,"customers":48},
    {"day":"Sat","sales":3000,"customers":62},
    {"day":"Sun","sales":2750,"customers":57},
]

def weekly_report(records):
  total_sales = 0
  total_customers = 0
  for r in records:
    total_sales+=r["sales"]
    total_customers+=r["customers"]

  average_day = total_sales/len(records)
  avg_basket = total_sales/total_customers

  best_day = records[0]
  weak_days = []
  for r in records:
    if r["sales"]>best_day["sales"]:
        best_day=r
    if r["sales"]<average_day:
        weak_days.append(r["day"])

  print(f"Total sales:  KES{total_sales:,}")
  print(f"Total customers:  {total_customers}")
  print(f"Average per day:  KES{average_day:,.2f}")
  print(f"Average basket size:  KES{avg_basket:,.2f}")
  print(f"Best day:   {best_day['day']} (KES{best_day['sales']:,})")
  print(f"Below-average days:{','.join(weak_days)}")

weekly_report(week)
```

### Mama Njeri's Business Review

The kiosk generated total sales of KES 13,230

**2. Customer Spending**

The average amount a customer spent is approximately KES 44.70.

**3. Best-Performing Day**

Saturday recorded the highest sales at KES 3,000.

**4. Days Requiring Improvement**

Monday, Tuesday, Wednesday and Thursday recorded sales below the weekly daily average of KES 1,890.
