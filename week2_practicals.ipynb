## Lab 2.1: Clean a Messy Survey Column

**Goal:** Combine list comprehensions and error handling to rescue usable values from dirty data.

```python

ages = ["23","31","unknown","45"," 29 ","-5","","38","200"]
print(ages)

def to_int_or_none(value):
    try:
        return int(value.strip())
    except ValueError:
        return None
print(to_int_or_none("45"))
print(to_int_or_none("unknown"))

converted_ages = [to_int_or_none(age) for age in ages]
print(converted_ages)

valid_ages = [
    age for age in converted_ages
    if age is not None and 15 <= age <= 100]
print(valid_ages)


# Count valid ages
valid_count = len(valid_ages)

# Count values that could not be converted
unparseable_count = sum(
    age is None for age in converted_ages
)

# Count converted values outside the acceptable age range
out_of_range_count = sum(
    age is not None and not (15 <= age <= 100)
    for age in converted_ages
)

print("Valid ages:", valid_count)
print("Unparseable values:", unparseable_count)
print("Out-of-range ages:", out_of_range_count)
```

### Why does 200 appear in converted_ages but not in valid_ages?

The value 200 is successfully converted from a string into an integer, so it appears in converted_ages. However, it falls outside the acceptable age range of 15 to 100 (15 <= age <= 100). Therefore, it is excluded from valid_ages.

This demonstrates the difference between data conversion and data validation.

## Lab 2.2 Write and Read a Matatu Fares File

**Goal:** Practise file input/output with correct type conversion.

```python
# Import CSV module
import csv

# Create the matatu fares dataset
routes = [
    {"route": "Nyeri-Nairobi", "fare": 700, "distance_km": 150},
    {"route": "Nyeri-Karatina", "fare": 150, "distance_km": 30},
    {"route": "Nyeri-Nanyuki", "fare": 300, "distance_km": 60},
    {"route": "Nyeri-Nakuru", "fare": 800, "distance_km": 162},
    {"route": "Nairobi-Nakuru", "fare": 600, "distance_km": 160},
    {"route": "Nairobi-Mombasa", "fare": 2000, "distance_km": 480}
]

print(routes)


# Write the dataset into a CSV file
with open("routes.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file, fieldnames=["route", "fare", "distance_km"]
    )

    writer.writeheader()
    writer.writerows(routes)
print("Routes successfully saved to routes.csv")

with open("routes.csv", "r") as file:
    print(file.read())


# Read the CSV file using csv.DictReader
with open("routes.csv", "r") as file:
    reader = csv.DictReader(file)
    loaded_routes = list(reader)
print(loaded_routes)


# Convert fare and distance_km into floating-point numbers
for route in loaded_routes:
    route["fare"] = float(route["fare"])
    route["distance_km"] = float(route["distance_km"])

print(loaded_routes)


# Calculate fare per kilometre
fare_per_km = [
    {
        "route": route["route"],
        "fare_per_km": route["fare"] / route["distance_km"]
    }
    for route in loaded_routes
]

for item in fare_per_km:
    print(
        f'{item["route"]}: KES {item["fare_per_km"]:.2f}/km'
    )


# Identify the best-value route
best_route = min(fare_per_km, key=lambda item: item["fare_per_km"])

print("Best-value route:", best_route["route"])
print(f'Fare per km: KES {best_route["fare_per_km"]:.2f}')
```

### Analysis and Conclusion

The analysis compared the cost of travelling per kilometre across six matatu routes.

The Nairobi–Nakuru route offers the best value, with a fare of KES 3.75 per kilometre. The Nyeri–Karatina and Nyeri–Nanyuki routes have the highest cost per kilometre at KES 5.00.

The analysis demonstrates how CSV file handling and Python calculations can transform raw transport data into useful information for comparing travel costs and making informed decisions.

**Conclusion:** Nairobi–Nakuru is the most cost-efficient route among the six routes analysed, based on fare per kilometre.

## Lab 2.3: NumPy Warm-Up

**Goal:** Build flluency with arrays, masks and aggregations.

```python
# Import NumPy
import numpy as np

# Generate 30 daily temperatures
rng = np.random.default_rng(1)
temperatures = rng.normal(24, 3, 30).round(1)
print(temperatures)


# Calculate descriptive statistics
mean_temp = np.mean(temperatures)
median_temp = np.median(temperatures)
std_temp = np.std(temperatures)
min_temp = np.min(temperatures)
max_temp = np.max(temperatures)

print(f"Mean temperature: {mean_temp:.2f} °C")
print(f"Median temperature: {median_temp:.2f} °C")
print(f"Standard deviation: {std_temp:.2f} °C")
print(f"Minimum temperature: {min_temp:.1f} °C")
print(f"Maximum temperature: {max_temp:.1f} °C")


# Count temperatures above 27°C using boolean mask and .sum()
hot_days = temperatures > 27
print(hot_days)

hot_day_count = hot_days.sum()
print("Days above 27°C:", hot_day_count)


# Convert Celsius to Fahrenheit using vectorised formula (c*9/5 + 32)
fahrenheit = temperatures * 9/5 + 32

print("Temperatures in Fahrenheit: ")
print(fahrenheit.round(1))


# Reshape the temperature array
weekly_temperatures = temperatures[:28].reshape(4, 7)
print("Weekly_temperatures: ")
print(weekly_temperatures)

# Calculate average temperature for each week
weekly_averages = weekly_temperatures.mean(axis=1)

print("Weekly average temperatures:")

for week, average in enumerate(weekly_averages, start=1):
    print(f"Week {week}: {average:.2f} °C")
```

### Findings

The generated dataset contains 30 daily temperatures, with a mean of 23.80°C and a median of 24.10°C. The standard deviation is 2.48°C, indicating relatively moderate variation.

Only one day recorded a temperature above 27°C. The minimum temperature was 15.9°C, while the maximum was 27.9°C.

After reshaping the first 28 observations into four weekly groups, Week 1 recorded the highest average temperature (24.43°C), while Week 4 recorded the lowest (22.64°C).

NumPy made it possible to perform these calculations efficiently using vectorised operations and array methods.
